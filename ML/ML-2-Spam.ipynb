## ML-2- Email classfication using KNN 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn import metrics
```

```python
df=pd.read_csv("emails.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
x=df.drop(["Email No.","Prediction"],axis=1)
y=df["Prediction"]
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python

#KNN
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
knn_pred=knn.predict(X_test)

#SVM
svm=SVC(kernel="linear")
svm.fit(X_train,y_train)
svm_pred=svm.predict(X_test)
```

```python
#Compare performance
knn_accuracy = metrics.accuracy_score(y_test,knn_pred)
svm_accuracy = metrics.accuracy_score(y_test,svm_pred)

if knn_accuracy>svm_accuracy:
    print("Knn has higher accuracy : ",knn_accuracy)
elif knn_accuracy<svm_accuracy:
    print("SVM has higher accuracy :",svm_accuracy)

else:
    print("Both has same accuracy :",knn_accuracy) 

```
